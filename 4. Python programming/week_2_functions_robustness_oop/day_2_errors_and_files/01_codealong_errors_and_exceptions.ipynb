# Errors and exceptions

**Goal:** stop programs from crashing on bad data, and fail with clear messages when something is really wrong.

In week 1 you learned to *read* error messages. Now you learn to *handle* them.

## 1. Recap: common exceptions

|---|---|
| `ValueError` | right type, wrong value: `int("abc")` |
| `TypeError` | wrong type: `"5" + 5` |
| `KeyError` | missing dictionary key |
| `IndexError` | list index out of range |
| `ZeroDivisionError` | dividing by zero |

## 2. `try` / `except`

Put the risky code in `try`. If an exception happens, Python jumps to `except` instead of crashing.

```python
text = "twelve"

try:
    number = int(text)
    print("Converted:", number)
except ValueError:
    print(f"Could not convert '{text}' to a number")

print("The program keeps running")
```

## 3. Catch specific exceptions

Name the exception you expect. A bare `except:` hides real bugs, so avoid it.

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None
    except TypeError:
        print("Both values must be numbers")
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "2"))
```

Catch several exceptions with one handler by using a tuple, and read the message with `as`:

```python
for raw in ["42", "4.2", "abc", None]:
    try:
        print(int(raw))
    except (ValueError, TypeError) as e:
        print(f"Skipping '{raw}': {e}")
```

## 4. Optional: `else` and `finally`

You will not need these often, but you will see them in other people's code.

- `else`: runs only if **no** exception happened
- `finally`: runs **always** (good for clean-up)

```python
def parse_quantity(text):
    try:
        quantity = int(text)
    except ValueError:
        print(f"'{text}' is not a whole number")
        return None
    else:
        print(f"'{text}' parsed OK")
        return quantity
    finally:
        print("-- done with", text)

parse_quantity("3")
parse_quantity("three")
```

## 5. Raising your own exceptions

Sometimes *you* decide the data is wrong. Use `raise` with a clear message.

```python
def validate_price(price):
    if price is None:
        raise ValueError("price is missing")
    if price < 0:
        raise ValueError(f"price cannot be negative, got {price}")
    return price

for p in [100, -5, None]:
    try:
        print("OK:", validate_price(p))
    except ValueError as e:
        print("Invalid:", e)
```

## 6. A defensive helper for messy data

Real data contains `"N/A"`, empty strings and typos. A helper that never crashes keeps your pipeline running.

```python
def to_float(text, default=None):
    """Convert text to float. Return `default` if that is not possible."""
    try:
        return float(text)
    except (ValueError, TypeError):
        return default

messy = ["100", "250.5", "N/A", "", None, "1e3"]
cleaned = []
for x in messy:
    cleaned.append(to_float(x))

print(cleaned)
print("failed:", cleaned.count(None))
```

## 7. `assert`: checks for *you*, the developer

`assert condition, message` raises `AssertionError` if the condition is False.

- **Exceptions** (`raise`): bad *input* you must handle
- **Assertions**: things that must be true, otherwise your *code* has a bug. Great for quick tests.

```python
assert to_float("3.5") == 3.5
assert to_float("abc") is None
assert to_float("abc", default=0.0) == 0.0
print("All checks passed")
```

```python
# This cell raises an error on purpose
assert to_float("7") == 8, "expected 8"
```

## 8. The pattern for data work: collect problems, do not crash

If one row out of a million is bad, you do not want the whole job to stop. Keep the good rows, **record** the bad ones, and report at the end.

```python
rows = ["10", "20", "x", "40", "", "60"]

good = []
bad = []
for raw in rows:
    try:
        good.append(int(raw))
    except ValueError:
        bad.append(raw)

print("good:", good)
print("bad :", bad)
```
