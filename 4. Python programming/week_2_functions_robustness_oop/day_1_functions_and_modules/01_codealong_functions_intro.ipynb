# introducing the return statement

```python
def customer_greeting(name):
    
    return f"Dear {name.title()}, welcome to our amazing website!"
```

```python
customer_greeting("felix")
```

```python
welcome_message = customer_greeting("felix")
```

```python
welcome_message
```

```python
# Now, the welcome_message variable is not empty!
# This is becauase we are now returning info in our customer_greeting() function
```

# creating a mean function

Our mean function will:
- accept a list with numbers as the input
- return the mean

```python
my_list = [1, 3, 10]
```

```python
# sum / no. of elements to calculate mean
```

```python
sum(my_list)
```

```python
len(my_list)
```

```python
sum(my_list)/len(my_list)
```

```python
def my_mean_func(num_list):
    
    mean = sum(num_list) / len(num_list)
    
    return mean
```

```python
mean = my_mean_func(my_list)
```

```python
mean
```

```python
my_new_list = [1,5,6,11]
```

```python
mean = my_mean_func(my_new_list)
```

```python
mean
```

# returning multiple values

accept a list of numbers and return
- the smallest number (min)
- the largest number (max)

```python
def minmax(number_list: list):
    minimum = min(number_list)
    maximum = max(number_list)
    return minimum, maximum
```

```python
list_for_test = [5, 30, 1000, 3]
```

```python
min_value, max_value = minmax(list_for_test)
```

```python
print(f"Min value: {min_value}")
print(f"Max value: {max_value}")
```
