```python
# Lets create a zone map for all of our stalls

# Create placeholder to hold the zones with a list of the names
stall_map = {}

# Loop over all the stalls
for  stall in stalls:
    name = stall.get("name")
    zone = stall.get("zone") # For every stall, we want to check what zone it belongs to

    # Check if the key exists, if not, we need to create it
    if zone not in stall_map:
        stall_map[zone] = [name]
    else:
        stall_map[zone].append(name)  # Append the stall to the correct list
```

```python
# Count the number of items sold per stall
order_counts = {}

# Loop over every order
for order in orders:
    # For every order, we need to extract the stall_id and the items ordered
    stall_id = str(order.get("stall_id"))
    items = order.get("items")
    # Loop over every item
    for item_info in items:
        # For every item, get the quantity
        quantity = item_info.get("qty")

        # Update the order_counts with that quantity for that specific stall
        # Check if the stall_id exisits in the dict, of not, we create it.
        if stall_id not in order_counts:
            order_counts[stall_id] = quantity
        else:
            order_counts[stall_id] += quantity
```

```python
### Lets find all the vegans at the Food festival
# Placeholder for storing our data
vegans = []

# Loop over each visitor
for visitor in visitors:
    prefs = visitor.get("prefs")
    # For every visitor, we want to loop over their prefs
    for pref_key, pref_value in prefs.items():
        if pref_key == "vegan_only":
            vegan_status = pref_value
            # If a visitor is vegan, we want to append them to the list
            if vegan_status == True:
                vegans.append(visitor)
```
