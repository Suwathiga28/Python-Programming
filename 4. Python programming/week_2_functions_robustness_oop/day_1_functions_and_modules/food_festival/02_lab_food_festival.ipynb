<details>

</details>

<details>
<summary>💡 Hint 2 - Code Structure</summary>
Start with <code>total = 0</code>. Loop through <code>order["items"]</code>. For each order item, loop through menus to find the matching item name, then add <code>price * qty</code> to total.

<pre>
def calculate_order_total(order):
    total = 0
    for order_item in order["items"]:
        item_name = order_item["item"]
        qty = order_item["qty"]
        # Find price in menus
        for menu_item in menus:
            if menu_item["item"] == item_name:
                price = menu_item["price"]
                total += price * qty
                break
    return total

### Exercise 8: Visitor Recommendations

Write a function called `get_visitor_recommendations` that:
- Takes one parameter: `visitor_name` (a string like "Alice")
- Returns a list of menu item names that match ALL of the visitor's preferences:
  - Price must be less than or equal to their `max_price`
  - If `vegan_only` is True, the item must be vegan
  - Spicy level must be less than or equal to their `avoid_spicy_above`

<details>
<summary>💡 Hint 1 - General Approach</summary>
First find the visitor in the visitors list to get their preferences. Then loop through menus and check if each item matches all the criteria.
</details>

<details>
<summary>💡 Hint 2 - Code Structure</summary>
Find the visitor first. Then create an empty recommendations list. Loop through menus with multiple if conditions (price, vegan, spicy level) - all must be true to add the item.
</details>

<details>
<summary>💡 Hint 3 - More Specific</summary>
<pre>
def get_visitor_recommendations(visitor_name):
    # Find the visitor
    visitor_prefs = None
    for visitor in visitors:
        if visitor["name"] == visitor_name:
            visitor_prefs = visitor["prefs"]
            break
    
    recommendations = []
    for item in menus:
        # Check all conditions
        if item["price"] <= visitor_prefs["max_price"]:
            if item["spicy_level"] <= visitor_prefs["avoid_spicy_above"]:
                if not visitor_prefs["vegan_only"] or item["is_vegan"]:
                    recommendations.append(item["item"])
    
    return recommendations
</pre>

### Exercise 9: Most Popular Item

Write a function called `find_most_popular_item` that:
- Takes no parameters
- Returns the name of the menu item that was ordered the most (by total quantity across all orders)

<details>
<summary>💡 Hint 1 - General Approach</summary>
You need to count the total quantity ordered for each item across all orders. A dictionary is perfect for keeping track of counts!
</details>

<details>
<summary>💡 Hint 2 - Code Structure</summary>
Create an empty dict. Loop through orders, then loop through each order's items. Add the qty to the dict (using item name as key). Then find which item has the highest count.
</details>

<details>
<summary>💡 Hint 3 - More Specific</summary>
<pre>
def find_most_popular_item():
    item_counts = {}
    
    # Count all items
    for order in orders:
        for order_item in order["items"]:
            item_name = order_item["item"]
            qty = order_item["qty"]
            
            if item_name in item_counts:
                item_counts[item_name] += qty
            else:
                item_counts[item_name] = qty
    
    # Find max
    max_count = 0
    most_popular = ""
    for item_name, count in item_counts.items():
        if count > max_count:
            max_count = count
            most_popular = item_name
    
    return most_popular
</pre>
</details>

</details>

<details>
<summary>💡 Hint 2 - Code Structure</summary>
Create an empty result list. Loop through orders where time matches. For each order, loop through stalls to find the matching name. Count items with <code>len(order["items"])</code>. Create a new dict with the info and append it.
</details>

<details>
<summary>💡 Hint 3 - More Specific</summary>
<pre>
def get_orders_with_stall_names(time):
    result = []
    
    for order in orders:
        if order["time"] == time:
            # Find stall name
            stall_name = ""
            for stall in stalls:
                if stall["id"] == order["stall_id"]:

            entry = {

                "stall_name": stall_name,
