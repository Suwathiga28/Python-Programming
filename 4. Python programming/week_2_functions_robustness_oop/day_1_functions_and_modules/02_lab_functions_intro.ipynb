# Lab Exercises: Introduction to Functions in Python

**Objective**: Get comfortable with defining and calling functions in Python.

**Note!** After you have written each function, make sure that it works as intended!

___

Exercise 1: Hello World Function

Description: Write a function named say_hello that prints "Hello, World!" when called.

Exercise 2: Double a Number

Description: Create a function named double_number that takes a number as an argument and returns twice its value.

Exercise 3: Return pi

Description: Write a function called return_pi that simply returns the number 3.14

Exercise 4: Check Zero

Description: Create a function named check_zero that takes a number and prints "Zero" if the number is 0, otherwise it prints "Not Zero".

Exercise 5: Increment by One

Description: Write a function called increment that takes a number and returns the number plus one.

Exercise 6: Is Negative

Description: Create a function named is_negative that checks if a number is negative. The function should return True if the number is negative, False otherwise.

Exercise 7: Length of String

Description: Create a function named string_length that returns the length of a given string.

Hint: Use the len() function to determine the length of the string.

Exercise 8: Length of list

Description: Create a function named list_length that returns the length of a given list.

Hint: Use the built in len() function to determine the length of the list.

Exercise 9: Return the First Element

Description: Write a function called first_element that takes a list and returns the first element of that list.

Exercise 10: Return biggest number

Description: Write a function called biggest_number that takes a list with numbers and returns the biggest number of that list

Hint: Use the built in max() function to find the biggest number

# Bonus exercises

# Fahrenheit to Celsius

Write a function that converts Fahrenheit to Celsius.

The formula to convert a temperature from Fahrenheit (°F) to Celsius (°C) is:

$$
5/9 \times (F - 32)
$$

Example test cases:
- 32 should return 0
- 212 hould return 100
- 98.6 should return 37

# Compound interest

Write a function to calculate compound interest given principal, rate, and number of years.

The formula is:

$$
A = P \times (1 + R)^{T}
$$
where:

- 𝐴 is the amount of money accumulated after T years, including interest.
- 𝑃 is the principal amount (the initial sum of money).
- 𝑅 is the annual interest rate (as a decimal).
- 𝑇 is the time the money is invested for in years.

Test Cases:

- Test Case 1: Calculate the compound interest for a principal of 1000, an annual interest rate of 5% (0.05 as a decimal), and a time period of 3 years. Expected Output: Approximately 1157.63
- Test Case 2: Calculate the compound interest for a principal of 1500, an annual interest rate of 4% (0.04 as a decimal), and a time period of 6 years. Expected Output: Approximately 1897.98
- Test Case 3: Calculate the compound interest for a principal of 2000, an annual interest rate of 3% (0.03 as a decimal), and a time period of 8 years. Expected Output: Approximately 2533.54

# Calculate Median

Write a function to find the median of a list of numbers.

- The function should take one argument, a list of numbers.
- Use the built in sorted() function to sort the list inside your function.

Test Cases:

- Test Case 1: A list with an odd number of elements: [3, 1, 2] | Expected Output: 2
- Test Case 2: A list with an even number of elements: [3, 1, 2, 4] | Expected Output: 2.5
- Test Case 3: A list with multiple similar elements: [1, 2, 2, 2, 3, 4] | Expected Output: 2

# Calculate mean and median

Write a function that takes a list of numbers, calculates the mean and median, and returns these values along with the original list in a dictionary format.

Test Cases:

- Test Case 1: Input: [3, 5, 1, 4]. Expected Output: {'Original List': [3, 5, 1, 4], 'Mean': 3.25, 'Median': 3.5}
- Test Case 2: Input: [20, 10, 50, 40, 30]. Expected Output: {'Original List': [20, 10, 50, 40, 30], 'Mean': 30.0, 'Median': 30}
- Test Case 3: Input: [1, 8, 9, 6, 7, 3, 3]. Expected Output: {'Original List': [1, 8, 9, 6, 7, 3, 3], 'Mean': 5.2857, 'Median': 6}

Hint:
- You can use the median function you created in the previous question
