## 🔍 Investigation 6: Convert Swedish dates to datetime
**Difficulty:** ⭐⭐⭐⭐☆ (4/5)

**Your mission:** The `date_sold` column contains dates in Swedish format like "Såld 13 april 2018". Convert these to proper datetime objects.

**This is challenging because:**
- Dates are in Swedish ("Såld" = "Sold", month names like "april", "juni", etc.)
- Pandas doesn't recognize Swedish dates by default
- You need to convert Swedish month names to numbers

**Approach:**
1. Remove "Såld " prefix
2. Convert Swedish month names to numbers (create a helper function)
3. Use `pd.to_datetime()` to parse

<details>
<summary>💡 Hint 1: Helper function approach</summary>
Create a function that replaces Swedish months with numbers:
```python
def swe_month_to_num(date_str):
    if "januari" in date_str:
        return date_str.replace("januari", "1")
    elif "februari" in date_str:
        return date_str.replace("februari", "2")
    # ... continue for all 12 months
```
</details>

<details>
<summary>💡 Hint 2: Swedish months</summary>
Swedish months: januari, februari, mars, april, maj, juni, juli, augusti, september, oktober, november, december
</details>

<details>
<summary>💡 Hint 3: Step-by-step transformation</summary>
```python
# Step 1: Remove "Såld "
dates_clean = df["date_sold"].str.replace("Såld ", "")

# Step 2: Convert Swedish months to numbers
dates_numeric = dates_clean.apply(swe_month_to_num)

# Step 3: Parse to datetime
df["date_sold"] = pd.to_datetime(dates_numeric, dayfirst=True)
```
</details>

<details>
<summary>🌟 Bonus: Dictionary approach</summary>
More elegant solution using a dictionary:
```python
month_map = {
    "januari": "1", "februari": "2", "mars": "3",
    # ... etc
}

def swe_month_to_num(date_str):
    for swe, num in month_map.items():
        if swe in date_str:
            return date_str.replace(swe, num)
```
</details>
