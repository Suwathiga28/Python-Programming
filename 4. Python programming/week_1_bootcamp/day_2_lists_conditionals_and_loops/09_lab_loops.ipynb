## 1. List comprehension basics

You are given this list:

```python
numbers = [1, 2, 3, 4, 5]
```

Then we are using syntax called **list comprehension** to create a new list called `squared` that contains each number squared.

Spend some time trying to understand what is going on with the list comprehension syntax

````python
numbers = [1, 2, 3, 4, 5]

squared = [number**2 for number in numbers]

print(numbers)
print(squared)
````
