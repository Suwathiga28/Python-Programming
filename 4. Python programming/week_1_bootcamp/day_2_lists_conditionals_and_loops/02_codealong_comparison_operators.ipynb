# Comparison Operators 

Comparison operators allow us to compare variables and output a Boolean value (True or False)

## and

all conditions must evalate to True, else it evaluates to False

```python
1 == 1 and 2 == 2
```

```python
1 == 1 and 2 == 3
```

```python
# Mika is at an amusement park and want's to ride a rollercoaster with age and height restrictions

# Mikas age and height
mika_age = 11
mika_height = 133
```

```python
# min age: 10
# min height: 130 cm 
mika_age > 10 and mika_height > 130
```

```python
# min age: 10
# min height: 140 cm 
mika_age > 10 and mika_height > 140
```

## or

at least 1 condition must evalate to True, else it evaluates to False

```python
2 == 3 or 1 == 1
```

```python
2 == 3 or 1 == 2
```

```python
a_var = 1.5
```

```python
type(a_var) == int or type(a_var) == float
```

```python
another_var = "1.5"
type(another_var) == int or type(another_var) == float
```

## in

- character or pattern present in a string
- element in a list


```python
# characters in strings
```

```python
"h" in "hello"
```

```python
# case sensitive
"H" in "hello"
```

```python
# elements in list
```

```python
my_list = [1, 2, 3, "hello!"]
```

```python
1 in my_list
```

```python
0 in my_list
```

```python
"hello!" in my_list
```

## not

```python
"h" not in "hello"
```

```python
my_list
```
