# Loops

Loops let you repeat code automatically. Essential for processing data, cleaning rows, and automating tasks.

## for loop

A for loop iterates over a list, starting at the first element and moving through each item.

```python
names = ["nils", "felix", "sara", "olivia"]

for name in names:
    print(name)
```

```python
# break - to explore what is going on
for name in names:
    break
```

## Operations in loops

You can perform operations inside loops and store results in a new list.

```python
capitalized_names = []

for name in names:
    capitalized_name = name.capitalize()
    print(capitalized_name)
    capitalized_names.append(capitalized_name)
```

```python
print(capitalized_names)
```

## Currency converter
You have a list sek_prices of prices in Swedish kronor (SEK).

Create an empty list called euro_prices and fill it with the converted values in euros.

Use the exchange rate 1 EUR = 10.9 SEK.

```python
sek_prices = [120, 55, 89, 240]
exchange_rate = 10.9
```

```python
euro_prices = []
for price in sek_prices:
    print(price/exchange_rate)
```

```python
for price in sek_prices:
    euro_price = price/exchange_rate
    euro_prices.append(euro_price)
```

## Conditions in loops

You can use if/elif/else statements inside loops to make decisions for each item.

```python
inventory_counts = [5, 20, 0, 13, 8]

for count in inventory_counts:
    if count == 0:
        print(f"{count}: Out of stock")
    elif count < 10:
        print(f"{count}: Low stock")
    else:
        print(f"{count}: Sufficient stock")
```

## Counting in loops

You can use a counter variable to count how many items meet a certain condition.

```python
responses = ["yes", "no", "yes", "yes", "no", "yes"]
yes_count = 0

for response in responses:
    if response == "yes":
        yes_count = yes_count + 1  # Add 1 to the counter
        
print(f"Total yes responses: {yes_count}")
```

```python
# Same thing using += (shorthand for adding to a variable)
yes_count = 0

for response in responses:
    if response == "yes":
        yes_count += 1  # Shorthand for: yes_count = yes_count + 1
        
print(f"Total yes responses: {yes_count}")
```

```python
# Using conditions to filter items
inventory = [5, 20, 0, 13, 8, 2]
low_stock = []

for count in inventory:
    if count < 10:
        low_stock.append(count)

print(low_stock)
```

## break

The break keyword stops a loop immediately when a condition is met.

```python
# I am looking for the keys
drawer = ["paper", "pens", "keys", "scissors", "toothpick"]

for item in drawer:
    print(f"Checking {item}")
    if item == "keys":
        print("Found the keys!")
        break
```

## List comprehension

List comprehension is a cleaner way to create lists from loops - all in one line.

```python
# Traditional way with a loop
names = ["nils", "felix", "sara", "olivia"]

capitalized_names = []
for name in names:
    capitalized_name = name.capitalize()
    print(capitalized_name)
    capitalized_names.append(capitalized_name)
```

```python
# Capitalizing names with list comprehension
capitalized = [name.capitalize() for name in names]
print(capitalized)
```

```python
# List comprehension with a condition, to use when you want to filter items
inventory = [5, 20, 0, 13, 8, 2]
low_stock = [count for count in inventory if count < 10]
print(low_stock)
```

## Nested loops: a loop inside a loop

A loop can contain another loop. The **inner loop runs completely for every item of the outer loop**.

Example: print every seat in a small cinema with 3 rows and 5 seats per row.

*(More examples for self-study: `bonus_and_review/06_codealong_nested_loops_optional.ipynb`)*

```python
rows = ["A", "B", "C"]
seats = [1, 2, 3, 4, 5]

for row in rows:
    for seat in seats:
        print(f"{row}{seat}", end=" ")
    print()  # new line after each row

# The inner loop ran 5 times for each of the 3 rows: 3 x 5 = 15 seats
```
