```python
my_list[0]
```

## append()

append() is an example of a list method

append() is used to add elements to a list

```python
shopping_basket = []
```

## remove()

remove() is another example of a list method

remove() is used to remove an element from a list

```python
shopping_basket.remove("yellow socks")
```

```python
shopping_basket
```

## Checking membership with `in`

The `in` operator checks if an element exists in a list

```python
"red tee" in shopping_basket
```

```python
"yellow socks" not in shopping_basket
```

## Sorting

Lists can be sorted in different ways - we are using the built in sorted() function here

```python
my_list = [10, 55, 111, 6]
```

```python
sorted(my_list)
```

```python
my_list
```

```python
my_sorted_list = sorted(my_list)
```

```python
my_sorted_list
```

```python
# introducing parameters
# not specific to lists! 
# functions and methods have params
sorted(my_sorted_list, reverse=True)
```

## Bonus: nested lists (if time allows)
Lists inside of a list

Let's see how this works!

```python
my_list1 = [1, 2, 3, 4]
my_list2 = [5, 6, 7, 8]
my_list3 = [9, 10, 11, 12]
```

```python
matrix = [my_list1, my_list2, my_list3]
```

### Bonus mini challenge! Grab the number 10 from the `matrix` variable!

```python
# instead of breaking it up into steps where we create a new variable
# we can just do it directly on the matrix variable
# where we first "retrieve" the third list and then "retrieve" the second element from that list
matrix[2][1]
```
