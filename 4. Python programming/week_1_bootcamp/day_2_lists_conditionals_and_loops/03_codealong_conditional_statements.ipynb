# Conditional logic

The `if`, `elif` and `else` statements are used for decision making.

Decision making is required when we want to execute a code only if a certain condition is satisfied.

syntax:
```python
if expression1:
   statement(s)
   
elif expression2:
   statement(s)
   
elif expression3:
   statement(s)
   
else:
   statement(s)
```

````python
if 1 == 1:
    print("The statement is true")
````

````python
if 1 == "1":
    print("The statement is true")
# 1 == "1" evaluates to false because 1 is an integer and "1" is a string
````

````python
# the example is that we are checking through some drawers to find our keys
# in this first example there are no keys in any drawer, so we will execute the code under the "else" statement
drawer1 = ["paper", "pens", "scissors", "toothpick"]

drawer2 = ["a bunch of random clutter"]

drawer3 = ["a bunch of random clutter", "pens", "tape"]
````

````python
if "keys" in drawer1:
    print("I found the keys in drawer ---> 1!")
    
elif "keys" in drawer2:
    print("I found the keys in drawer ---> 2!")
    
elif "keys" in drawer3: 
    print("I found the keys in drawer ---> 3!")
    
else:
    print("The keys are nowhere to be found!")
````

````python
drawer3.append("keys")
# rerun code above...
# now the second elif statement evaluates to True, and the code inside it gets executed
````

# Nested conditional statements

Often you need to check more than one condition before deciding what to do.
A nested conditional statement means putting one if (or elif/else) inside another.
It lets your program make decisions within decisions - like checking a second rule only if the first one is true.

Simple yet powerful!

Mind the indentation!

````python
age = 17
has_ticket = True
````

````python
if has_ticket:
    if age >= 18:
        print("Welcome to the movie!")
    else:
        print("Sorry, you’re too young.")
else:
    print("You need a ticket to enter.")
````

````python
# "flatting" the nested conditions
# More efficient but less control
````

````python
if has_ticket and age >= 18:
    print("Welcome to the movie!")
else: 
    print("You need to have a ticket and be 18 yrs or older to enter.")
````
