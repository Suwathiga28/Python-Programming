# Bonus questions

For the bonus questions you will generate a dataset

````python
# Execute this cell to create a new "transactions" list with 100 dicts (and with dates)
import random

products = [
    ("Coffee Beans", "Grocery", 89),
    ("Apples", "Grocery", 5),
    ("Laptop Stand", "Electronics", 299),
    ("USB Cable", "Electronics", 79),
    ("Mouse", "Electronics", 199),
    ("Notebook", "Office", 25),
    ("Stapler", "Office", 59),
    ("T-Shirt", "Clothes", 149),
    ("Sweater", "Clothes", 349),
    ("Water Bottle", "Sports", 129),
]

transactions = []

for i in  range(1, 101):   # 100 transactions
    product, category, price = random.choice(products)
    quantity = random.randint(1, 5)
    customer_id = random.randint(100, 120)
    
    # random date between Jan 1–31 2024
    day = random.randint(1, 31)
    date = f"2025-10-{day:02d}"
    
    transactions.append({
        "transaction_id": i,
        "customer_id": customer_id,
        "product": product,
        "category": category,
        "quantity": quantity,
        "price": price,
        "date": date
    })
````
