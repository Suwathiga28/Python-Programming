## Changing and adding in dictionaries

- update()
- [] notation

```python
countries_and_capitals = {
    "sweden": "stockholm",
    "norway": "oslo",
    "denmark": "copenhagen"
}
```

```python
# we want to change to stockholm is capitalized
```

```python
print(countries_and_capitals)
countries_and_capitals.update({"sweden": "Stockholm"})
print(countries_and_capitals)
# we can see from the print statements what we change "stockholm" to "Stockholm" with a capitalized S
```

```python
# there is another way of updating a dict
# the [] notation
print(countries_and_capitals)
countries_and_capitals["norway"] = "Oslo"
print(countries_and_capitals)

# we can see from the print statements what we change "osla" to "Osla" with a capitalized O
```

## lists inside dictionaries

```python
# A contact with multiple phone numbers and emails
contact = {
    "name": "Sofia",
    "phones": ["555-1234", "555-5678"],
    "emails": ["sofia@work.com", "sofia@personal.com"]
}

print(contact)
```

```python
# retrieve "sofia@personal.com"
contact.get("emails")
# this returns a list
# we want the 2nd element of the list (index position 1)
```

```python
# so we combine .get() with indexing [1]
contact.get("emails")[1]
```

## Dictionaries inside Dictionaries

```python
spotify_users = {
    "alice99": {
        "first_name": "Alice",
        "last_name": "Lund",
        "email": "alice.lund@example.com",
        "subscription": "Premium"
    },
    "tommy_s": {
        "first_name": "Tommy",
        "last_name": "Svensson",
        "email": "tommy.s@example.com",
        "subscription": "Free"
    },
    "beatfan": {
        "first_name": "Sara",
        "last_name": "Berg",
        "email": "sara.berg@example.com",
        "subscription": "Premium"
    }
}
```

```python
# we want to find the subscription for the user "alice99"
```

```python
spotify_users.get("alice99").get("subscription")
# we can chain multiple .get() operations to go "deeper" into a nested dictionary
```

## keys(), values() and items() methods



```python
print(countries_and_capitals)
```

```python
# keys() is a dictionary method that returns all keys
countries_and_capitals.keys()
```

```python
# values() is a dictionary method that returns all values
countries_and_capitals.values()
```

```python
# items() is a dictionary method that returns all items (also known as key-value pairs)
countries_and_capitals.items()
```

## looping over a dictionary

```python
for key in countries_and_capitals:
    print(key)
```

```python
for key, value in countries_and_capitals.items():
    print(key, value)
```

```python
for country, capital in countries_and_capitals.items():
    print(country, capital)
# compare readability!
# Now instead of key, value we name the variables country, capital
# Much easier to read!
```
