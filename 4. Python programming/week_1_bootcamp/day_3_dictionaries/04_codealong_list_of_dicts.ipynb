```python
shoes = [
    # Running
    {"name": "Pegasus 40", "brand": "Nike", "category": "Running", "year": 2025, "price": 1390, "discount": 0},
    {"name": "Ultraboost Light", "brand": "Adidas", "category": "Running", "year": 2024, "price": 1790, "discount": 0},
    {"name": "FuelCell Rebel v3", "brand": "New Balance", "category": "Running", "year": 2023, "price": 1590, "discount": 0},
    {"name": "Gel-Nimbus 25", "brand": "Asics", "category": "Running", "year": 2024, "price": 1990, "discount": 0},

    # Trail
    {"name": "Speedcross 6", "brand": "Salomon", "category": "Trail", "year": 2025, "price": 1490, "discount": 0},
    {"name": "Pegasus Trail 4", "brand": "Nike", "category": "Trail", "year": 2023, "price": 1390, "discount": 0},
    {"name": "Terrex Agravic", "brand": "Adidas", "category": "Trail", "year": 2024, "price": 1290, "discount": 0},
    {"name": "Cascadia 17", "brand": "Brooks", "category": "Trail", "year": 2023, "price": 1590, "discount": 0},

    # Training
    {"name": "Metcon 8", "brand": "Nike", "category": "Training", "year": 2024, "price": 1290, "discount": 0},
    {"name": "Nano X3", "brand": "Reebok", "category": "Training", "year": 2024, "price": 1390, "discount": 0},
    {"name": "Dropset Trainer", "brand": "Adidas", "category": "Training", "year": 2023, "price": 1200, "discount": 0},
    {"name": "TRX Trainer", "brand": "Under Armour", "category": "Training", "year": 2024, "price": 1100, "discount": 0},
]
```

```python
len(shoes)
# 12 dicts inside our list
```

```python
# inspect dict at index pos 0
shoes[0]
```

```python
# looping over the list of dicts
for shoe_dict in shoes:
    print(shoe_dict)
    print("**"*50)
```

```python
# display product name and price for each product
```

```python
for shoe_dict in shoes:
    name = shoe_dict.get("name")
    price = shoe_dict.get("price")
    
    print(f"Product name: {name}, price: {price}")
    print("**"*20)
```

```python
# display product name and price if brand == Nike
```

```python
for shoe_dict in shoes:
    brand = shoe_dict.get("brand")
    name = shoe_dict.get("name")
    price = shoe_dict.get("price")
    
    if brand == "Nike":
        print(f"Product name: {name}, price: {price}")
        print("**"*50)
```

```python
# Nike is running a campaign on all shoes (except releases from year 2025) 
# discount for 2024 is 10% 
# discount for 2023 is 15%
```

```python
# setting new values for discount
for shoe_dict in shoes: 
    if shoe_dict.get("brand") == "Nike":
        if shoe_dict.get("year") == 2024:
            shoe_dict["discount"] = 0.1
        elif shoe_dict.get("year") == 2023:
            shoe_dict["discount"] = 0.15
```

```python
# display the new discounted prices
```

```python
for shoe_dict in shoes:
    brand = shoe_dict.get("brand")
    name = shoe_dict.get("name")
    price = shoe_dict.get("price")
    discount = shoe_dict.get("discount")
    discounted_price = price * (1-discount)
    # round to nearest 0 or 5
    # final_price = round(discounted_price / 5) * 5
    
    if brand == "Nike":
        print(f"Product name: {name}, price: {discounted_price}")
        print("**"*50)
```
