# Nested Loops

Nested loops are loops inside other loops. They help when you need to combine or pair items from different lists.

Key concept: The inner loop completes all its iterations before the outer loop moves to the next item.

## Cinema seats

Print every seat in a small cinema with 3 rows and 5 seats per row.

```python
rows = ["A", "B", "C"]       
seats = [1, 2, 3, 4, 5]         

for row in rows:
    for seat in seats:
        print(f"{row}{seat}")
    print()
```

```python
# Let's see what's happening
for row in rows:
    print(f"Starting row {row}")
    for seat in seats:
        print(f"  Seat {seat}")
```

```python
# How many times does this run?
count = 0

for row in rows:
    for seat in seats:
        count += 1
        print(f"Iteration {count}: {row}{seat}")

print(f"\nTotal: {count} iterations")
print(f"This is 3 rows × 5 seats = 15")
```

## Storing results in a list

```python
rows = ["A", "B", "C"]       
seats = [1, 2, 3, 4, 5]    

seat_map = []

for row in rows:
    for seat in seats:
        seat_map.append(f"{row}{seat}")

seat_map
```

```python
# Creating a 2D list (list of lists)
rows = ["A", "B", "C"]       
seats = [1, 2, 3, 4, 5]    

seat_map = []

for row in rows:
    row_seats = []
    for seat in seats:
        row_seats.append(f"{row}{seat}")
    seat_map.append(row_seats)

seat_map
```

## Multiplication table

Another example with numbers.

```python
# Small multiplication table
numbers = [1, 2, 3, 4]

for i in numbers:
    for j in numbers:
        result = i * j
        print(f"{i} × {j} = {result:2}" , end="  ")  # :2 formats to 2 spaces
    print()  # New line after each row
```

## Menu combinations

Combining products with sizes.

```python
drinks = ["coffee", "tea", "juice"]
sizes = ["small", "medium", "large"]

for drink in drinks:
    for size in sizes:
        print(f"{size} {drink}")
```

```python
# Count specific combinations
count = 0

for drink in drinks:
    for size in sizes:
        if size == "large":
            count += 1
            print(f"{size} {drink}")

print(f"\nTotal large drinks: {count}")
```
