# Part 2: Sales employees dataset

In part 2 of the merge lab we will work with a new dataset. This dataset stores information related to sales employees in a company store.

There are 7 tables in total (stored in the `sales_emp_dataset` folder):

- **4 section tables**: Employee IDs for sales employees working in sections a, b, c & d
- **1 salaries table**: Monthly salaries for each sales employee
- **1 sales table**: Total sales in SEK in March for each sales employee
- **1 bonus factors table**: Bonus factors for employees in the bonus program

**Understanding bonuses:** If employee X has made sales of 100,000 SEK in March and their bonus factor is 0.01, they will receive a bonus of 1,000 SEK (100,000 × 0.01) in March. Not all employees are included in the bonus program.
