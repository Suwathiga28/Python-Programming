### Decision flowchart

**Question: How should I combine my DataFrames?**

1. **Do the DataFrames have the same columns?**
   - YES → Use `pd.concat()` to stack rows
   - NO → Continue to question 2

2. **Do you want to match rows based on common values?**
   - YES → Use `.merge()` with appropriate `how` parameter
   - NO → Use `pd.concat(axis=1)` to place side-by-side (rare)

3. **What type of join do you need?**
   - Only matching data → `how="inner"`
   - All data from both → `how="outer"`
   - Keep all from primary table → `how="left"`

## 10. Summary

### Combining DataFrames: Three approaches

✅ **1. Concatenation (`pd.concat()`)**
- **Purpose**: Stack or place DataFrames side-by-side
- **Use for**: 
  - Adding rows from similar datasets (e.g., data from different time periods)
  - Adding columns side-by-side (less common)
- **Key parameters**:
  - `axis=0` (default): Stack rows vertically
  - `axis=1`: Place columns horizontally
  - `ignore_index=True`: Create new sequential index
  - `keys=[...]`: Add multi-level index to track sources

✅ **2. Merging (`DataFrame.merge()`)**
- **Purpose**: Intelligently combine DataFrames based on matching values
- **Use for**: Joining related data (like SQL JOINs)
- **Join types**:
  - `how="inner"` (default): Only matching rows
  - `how="outer"`: All rows from both DataFrames
  - `how="left"`: All rows from left, matching from right
  - `how="right"`: All rows from right, matching from left
- **Key parameters**:
  - `on="column"`: Column(s) to match on
  - `left_on`, `right_on`: Different column names
  - `left_index`, `right_index`: Join on indexes
  - `suffixes=("_x", "_y")`: Handle duplicate column names
  - `indicator=True`: Track merge sources

✅ **3. Joining (`DataFrame.join()`)**
- **Purpose**: Simplified index-based joins
- **Use for**: Quick index-based combinations
- **Note**: `.merge()` is generally more flexible

⚠️ **Notice**: 
- `.join()` uses `lsuffix` and `rsuffix` (not `suffixes`)
- `.join()` defaults to `how="left"` (not `how="inner"` like `.merge()`)
- `.join()` is shorter for index-based joins but less explicit

✅ **Key Takeaway**: Stick with `.merge()` for clarity and consistency!

```python
# Using join() - automatically joins on index
result_join = men2004_indexed.join(
    men2008_indexed,
    lsuffix="_2004",
    rsuffix="_2008"
)
result_join.head()
```

```python
# Using merge() with indexes
result_merge = men2004_indexed.merge(
    men2008_indexed,
    left_index=True,
    right_index=True,
    suffixes=("_2004", "_2008")
)
result_merge.head()
```

### 4.3 Adding a multi-level index with keys

```python
# We can drop the old index column if we don't need it
men0408_clean = men0408_all_reset.drop(columns=["Index"])
men0408_clean
```

## 6. Introduction to merging and joining

📚 **Concept**: **Merging** (also called **joining**) combines DataFrames based on common values in specified columns, not just by position.

Think of it like matching puzzle pieces - you look for matching patterns (common values) to connect the right pieces together.

**Key difference from concatenation:**
- **Concatenation**: Combines by position (stack or place side-by-side)
- **Merging**: Combines by matching values in specific columns

**Common use cases:**
- Matching customers to their orders (by customer ID)
- Combining student grades with student demographics (by student ID)
- Linking products to categories (by category ID)

<!-- INSTRUCTOR: Emphasize this is similar to SQL JOINs for students with database experience -->

### 6.1 Simple merge example

```python
# Reload data
men2004 = pd.read_csv("men2004.csv")
men2008 = pd.read_csv("men2008.csv")
```

```python
# Merge based on the "Athlete" column
merged = men2004.merge(men2008, on="Athlete")
merged
```

💡 **What happened?**
- Pandas looked for athletes who appear in **both** DataFrames
- Only matching athletes are included
- Columns with the same name get suffixes: `_x` (from left DataFrame) and `_y` (from right DataFrame)

```python
# We can customize the suffixes
merged = men2004.merge(men2008, on="Athlete", suffixes=("_2004", "_2008"))
merged
```

✅ **Key Takeaway**: By default, `.merge()` performs an **inner join** - it only keeps rows where the key exists in **both** DataFrames.
