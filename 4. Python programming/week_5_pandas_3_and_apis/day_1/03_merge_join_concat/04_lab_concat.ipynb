# Concat lab

The main purpose of this lab is for you to practice concatenating DataFrames. You will also need to apply other pandas skills to solve these exercises.

## Dataset

In this lab we will use CSV files with information about apartments sold on Bondegatan. Each CSV file stores information about apartments sold during one year, from 2013 up to 2020.

**Task:** Put the information about apartments sold on Bondegatan in the years 2013, 2014 & 2015 in the same DataFrame.

<details>
<summary>💡 Hint</summary>
Use pd.concat() with a list of the three DataFrames. Consider using the keys parameter to track which year each row came from.
</details>

## Exercise 2: Handle column name mismatches
**Difficulty:** ⭐⭐⭐☆☆ (3/5)

**Task:** Put the information about apartments sold on Bondegatan in the years 2019 & 2020 in the same DataFrame.

**Challenge:** When you try to concatenate these files, you'll notice a problem! The column names for 2019 are in English and the column names for 2020 are in Swedish, so the resulting DataFrame will have twice the amount of columns and lots of NaN values.

Your task is to fix this by changing the column names in the 2020 data from Swedish to English before concatenating.

**Note:** The columns are in the same order, so you don't need to know the translations - you can copy the column names from the 2019 DataFrame.

<details>
<summary>💡 Hint</summary>
Load the 2019 data first, then load the 2020 data and use .columns assignment or .rename() to match the 2019 column names before concatenating.
</details>
