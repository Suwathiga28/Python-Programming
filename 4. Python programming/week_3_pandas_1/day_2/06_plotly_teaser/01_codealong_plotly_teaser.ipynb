# Plotly teaser (about 1 hour)

You already know pandas. Plotly Express turns a DataFrame into an interactive chart in **one line**.

Install once: `pip install plotly nbformat`

```python
import pandas as pd
import plotly.express as px

df = px.data.gapminder()
df.head()
```

## Bar chart

```python
latest = df[df["year"] == 2007]
by_continent = latest.groupby("continent", as_index=False)["pop"].sum()
px.bar(by_continent, x="continent", y="pop", title="Population by continent, 2007")
```

## Line chart

```python
sweden = df[df["country"].isin(["Sweden", "Norway", "Denmark"])]
px.line(sweden, x="year", y="lifeExp", color="country", title="Life expectancy")
```

## Histogram

```python
px.histogram(latest, x="lifeExp", nbins=20, title="Life expectancy distribution, 2007")
```

**Try it:** hover over the charts, zoom, click the legend. Then change `color=`, `title=` and the data. The full Plotly toolbox comes in week 6.
