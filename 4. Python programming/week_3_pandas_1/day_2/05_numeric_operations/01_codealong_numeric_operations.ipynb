## 1. Introduction to numeric operations

📚 **Concept**: One of pandas' most powerful features is **vectorization** - the ability to perform operations on entire columns of data at once, without writing loops.

Think of it like this:
- Traditional programming: Loop through each row, perform calculation
- Pandas: Apply calculation to entire column in one operation

This makes pandas code:
- ✅ Faster (optimized under the hood)
- ✅ More readable (less code to write)
- ✅ Less error-prone (no loop logic to debug)

## 2. Vectorized arithmetic operations

📚 **Concept**: You can perform arithmetic operations on entire columns using standard operators: `+`, `-`, `*`, `/`, `**` (exponentiation)

These operations apply to **every value** in the column automatically!

<!-- INSTRUCTOR: Demonstrate simple operations first, then build to more complex calculations -->

### 2.1 Simple arithmetic example

Let's calculate the year of birth for passengers. The Titanic sank in 1912.

````python
# View some ages
df["age"].head(10)
````

````python
# Calculate year of birth: 1912 minus age
df["year_of_birth"] = 1912 - df["age"]
df[["age", "year_of_birth"]].head(10)
````

✅ **Key Takeaway**: The calculation `1912 - df["age"]` applied to **all 891 passengers** in one operation, no loop needed!

💡 **Note**: Missing values (NaN) in the age column result in NaN for year_of_birth - pandas handles this automatically.

### 2.2 Multi-step calculations

📚 **Concept**: You can combine multiple operations to perform complex calculations. Let's convert the historical fare prices to modern Swedish kronor (SEK).

We need to:
1. Account for inflation from 1912 to 2021
2. Convert from USD to SEK

<!-- INSTRUCTOR: Break down the calculation step by step, explain each component -->

````python
# Define our conversion factors
usd_to_sek = 8.8  # Exchange rate
inflation_rate = 1.05  # 5% annual inflation
years_elapsed = 2021 - 1912  # Number of years

print(f"Years elapsed: {years_elapsed}")
print(f"Total inflation multiplier: {inflation_rate**years_elapsed:.2f}")
````

💡 **Understanding the calculation**:
- `inflation_rate**years_elapsed` uses the exponent operator (`**`) to calculate compound inflation
- After 109 years at 5% annual inflation, prices are about 200 times higher!

````python
# Calculate fare in 2021 SEK
df["fare_sek_2021"] = df["fare"] * (inflation_rate ** years_elapsed) * usd_to_sek

# View original and converted fares
df[["fare", "fare_sek_2021"]].head()
````

### 2.3 Exploring the results

Let's find the most and least expensive tickets:

````python
# Top 5 most expensive tickets
df["fare_sek_2021"].nlargest(5)
````

💡 **Interpretation**: The most expensive tickets cost nearly 1 million SEK in today's money - first class on the Titanic was very expensive!

````python
# Bottom 5 cheapest tickets
df["fare_sek_2021"].nsmallest(5)
````

⚠️ **Note**: Zero-value fares likely indicate staff, children traveling free, or data issues.

## 4. Aggregation methods

📚 **Concept**: **Aggregation** means combining multiple values into a single summary value. Common aggregations:

- `.sum()` - Add all values
- `.mean()` - Calculate average
- `.min()` - Find minimum
- `.max()` - Find maximum
- `.median()` - Find middle value
- `.std()` - Calculate standard deviation

## 5. The axis parameter explained

📚 **Concept**: The **`axis` parameter** controls the direction of an operation:

- `axis=0` (default): Operate down columns (combine rows) → Column-wise
- `axis=1`: Operate across rows (combine columns) → Row-wise

**Visual guide**:
```
       Mon  Tue  Wed  Thu  Fri
Steven  34   27   15  NaN   33
Mike    45    9   74   87   12
       
axis=0 ↓    ↓    ↓    ↓    ↓   (sum down each column)
axis=1 →  →  →  →  →           (sum across each row)
```

💡 **Memory aid**: Think of axis=0 as "zero rows remaining" (you collapse rows) and axis=1 as "one row per person" (you collapse columns)

<!-- INSTRUCTOR: Use the visual and have students predict results before running code -->

### 5.1 axis=0: Column-wise operations (default)

````python
# Sum down each column (total sales per day)
sales.sum(axis=0)
````

### 5.2 axis=1: Row-wise operations

````python
# Sum across each row (total sales per person for the week)
sales.sum(axis=1)
````

✅ **Key Takeaway**: 
- axis=0 → Result has one value per column
- axis=1 → Result has one value per row

### 5.3 Creating a totals column

Let's add a column showing each person's weekly total:

````python
# Create new column with weekly totals for each person
sales["weekly_total"] = sales.sum(axis=1)
sales
````

⚠️ **Note**: NaN (missing) values are ignored in sum calculations by default. Steven's total is 109 (not including Thursday's NaN).

## 7. Working with datetime and stock data

📚 **Concept**: Financial data often involves dates and calculating returns over time. Let's explore stock portfolio analysis.

<!-- INSTRUCTOR: Simplify the stock example - focus on core concepts not complex operations -->

### 7.1 Loading stock data

````python
# Load historical stock prices
stocks = pd.read_csv("stocks.csv")
stocks.head()
````

````python
# Check the data structure
stocks.info()
````

⚠️ **Notice**: The `date` column is stored as `object` (text), not as a date type!

### 7.2 Converting to datetime

📚 **Concept**: Pandas has a special `datetime64` type for dates. Converting text to datetime enables date-specific operations.

````python
# Convert date column to datetime type
stocks["date"] = pd.to_datetime(stocks["date"])
````

````python
# Verify the conversion
stocks.info()
````

✅ **Success!** Now the date column is `datetime64[ns]` type, enabling date arithmetic and filtering.

### 7.3 Exploring the data

````python
# How many unique stocks?
num_stocks = stocks["Name"].nunique()
print(f"Number of unique stocks: {num_stocks}")
````

````python
# View a sample of stock names
stocks["Name"].unique()[:20]  # First 20 stock tickers
````

### 7.4 Simple portfolio analysis

Let's select 5 random stocks and calculate portfolio value over time:

````python
# Select 5 random stocks for our portfolio
import random
random.seed(42)  # For reproducibility

all_stocks = stocks["Name"].unique()
portfolio_stocks = random.sample(list(all_stocks), 5)
print(f"Our portfolio: {portfolio_stocks}")
````

````python
# Filter data for only our portfolio stocks
portfolio = stocks[stocks["Name"].isin(portfolio_stocks)]
portfolio.head()
````

### 7.5 Calculating portfolio value

📚 **Concept**: To calculate total portfolio value, we sum the prices of all stocks on each date.

````python
# Reshape data: dates as rows, stocks as columns
portfolio_wide = portfolio.pivot(index="date", columns="Name", values="close")
portfolio_wide.head()
````

````python
# Calculate total portfolio value (sum across stocks for each date)
portfolio_wide["total_value"] = portfolio_wide.sum(axis=1)
portfolio_wide[["total_value"]].head()
````

### 7.6 Calculating return on investment (ROI)

📚 **Concept**: ROI measures investment performance:

**Formula**: `(Final Value / Initial Value) * 100`

Example: If you invest $100 and it becomes $150, ROI = (150/100) * 100 = 150% (a 50% gain)

````python
# Get first and last values
initial_value = portfolio_wide["total_value"].iloc[0]
final_value = portfolio_wide["total_value"].iloc[-1]

print(f"Initial portfolio value: ${initial_value:.2f}")
print(f"Final portfolio value: ${final_value:.2f}")
````

````python
# Calculate ROI percentage
roi = (final_value / initial_value) * 100
gain_percent = roi - 100

print(f"ROI: {roi:.2f}%")
print(f"Total gain: {gain_percent:.2f}%")
````

💡 **Interpretation**: 
- ROI of ~193% means the portfolio is worth 1.93x the initial investment
- This represents a ~93% gain over the time period
- Of course, this is just one random sample - results vary!
