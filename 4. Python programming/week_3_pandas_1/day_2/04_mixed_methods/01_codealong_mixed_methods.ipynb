## 1. Data inspection methodsBefore analyzing any dataset, we need to understand its structure and content. Pandas provides several powerful inspection methods that help us quickly grasp what we're working with.

### 1.1 Loading and initial exploration

````python
import pandas as pdimport numpy as np# Load the Titanic datasettitanic = pd.read_csv("titanic.csv")
````

📚 **Concept: The shape attribute**The `.shape` attribute gives us the dimensions of our DataFrame as a tuple: (rows, columns). This is always the first thing to check when loading data.

````python
# Check the size of our datasetprint(f"Rows: {titanic.shape[0]}")print(f"Columns: {titanic.shape[1]}")
````

### 1.2 Viewing data samples

📚 **Concept: head() and tail()**These methods let us peek at the data:- `.head(n)` shows the first n rows (default is 5)- `.tail(n)` shows the last n rows (default is 5)This is useful for understanding data structure without loading the entire DataFrame into memory.

````python
# View first 3 rowstitanic.head(3)
````

### 1.3 Understanding data types and structure

### 1.4 Statistical summaries

## 2. Column operationsWorking with column names is a fundamental skill. We often need to rename columns for clarity, select specific columns, or reorder them.

### 2.1 Renaming columns

📚 **Concept: The rename() method**The `.rename()` method allows us to change column (or index) names. We pass a dictionary where keys are old names and values are new names.**Modern best practice:** Assign the result to a variable rather than using `inplace=True`.

````python
# First, let's check current column namesprint(titanic.columns.tolist())
````

````python
# Rename specific columns for claritytitanic = titanic.rename(columns={    "pclass": "passenger_class",    "sibsp": "siblings_spouses",    "parch": "parents_children"})titanic.columns.tolist()
````

💡 **Tip:** You don't need to include all columns in the dictionary—only the ones you want to change.

⚠️ **Warning:** Avoid using `inplace=True` in modern pandas code:```python# Old way (discouraged)titanic.rename(columns={...}, inplace=True)# Modern way (preferred)titanic = titanic.rename(columns={...})```The assignment pattern makes code more readable and easier to debug.

### 2.2 Selecting and reordering columns

📚 **Concept: Column selection with double brackets**Using `df[[col1, col2, ...]]` we can:- Select specific columns- Reorder columns- Create a new DataFrame with only the columns we need

````python
# Select only demographic columnsdemographics = titanic[["name", "age", "sex", "passenger_class"]]demographics.head()
````

````python
# Reorder: put survived firsttitanic_reordered = titanic[[    "survived", "name", "passenger_class", "age", "sex",     "siblings_spouses", "parents_children", "fare", "embarked"]]titanic_reordered.head()
````

✅ **Key Takeaway:** Column operations are non-destructive by default. Always assign results to a variable to preserve changes.

## 3. Sorting and rankingSorting data helps us find extremes, identify patterns, and prepare data for analysis. Pandas offers several powerful sorting methods.

### 3.1 Sorting by values

📚 **Concept: The sort_values() method**`.sort_values()` sorts a DataFrame by one or more columns:- `by`: column name(s) to sort by- `ascending`: True for smallest-to-largest (default), False for largest-to-smallest- For multiple columns: pass a list to `by` and `ascending`

````python
# Sort by age (youngest to oldest)titanic_by_age = titanic.sort_values(by="age")titanic_by_age.head()
````

````python
# Sort by age (oldest to youngest)titanic_oldest = titanic.sort_values(by="age", ascending=False)titanic_oldest.head()
````

````python
# Sort by multiple columns: class ascending, then fare descendingtitanic_multi = titanic.sort_values(    by=["passenger_class", "fare"],    ascending=[True, False])titanic_multi.head(10)
````

💡 **Tip:** Multi-column sorting is great for finding patterns like "highest-paying passengers in each class".

### 3.2 Finding top and bottom values

📚 **Concept: nlargest() and nsmallest()**These methods provide shortcuts for finding extremes:- `.nlargest(n, column)` returns the top n rows by column value- `.nsmallest(n, column)` returns the bottom n rows by column valueThey're more efficient and readable than sorting the entire DataFrame.

````python
# Top 5 highest farestitanic.nlargest(5, "fare")[["name", "fare", "passenger_class"]]
````

````python
# Top 5 oldest passengerstitanic.nlargest(5, "age")[["name", "age", "survived"]]
````

````python
# 5 youngest passengerstitanic.nsmallest(5, "age")[["name", "age", "passenger_class"]]
````

⚠️ **Warning:** NaN (missing) values are automatically excluded from nlargest() and nsmallest() results.

✅ **Key Takeaway:** Use nlargest()/nsmallest() for quick insights into extreme values without sorting the entire dataset.

## 4. Value explorationUnderstanding the distribution and variety of values in your data is crucial. These methods help you explore categorical and discrete data.

### 4.1 Counting value frequencies

📚 **Concept: The value_counts() method**`.value_counts()` counts the frequency of each unique value in a Series:- Returns results sorted by frequency (most common first)- Can show proportions with `normalize=True`- Excludes NaN by default (use `dropna=False` to include)

````python
# Count survivors vs non-survivorstitanic["survived"].value_counts()
````

````python
# Show as proportionstitanic["survived"].value_counts(normalize=True)
````

````python
# Count passengers by classtitanic["passenger_class"].value_counts()
````

````python
# Count embarkation ports (including missing values)titanic["embarked"].value_counts(dropna=False)
````

💡 **Tip:** value_counts() is perfect for exploring categorical variables and finding data quality issues like unexpected categories or imbalanced classes.

### 4.2 Finding unique values

📚 **Concept: unique() and nunique()**Two related methods for understanding value variety:- `.unique()` returns an array of all distinct values- `.nunique()` returns the count of distinct values (faster, more memory-efficient)

````python
# See all unique values in embarked columnprint(titanic["embarked"].unique())
````

````python
# Count how many unique valuesprint(f"Unique embarkation ports: {titanic['embarked'].nunique()}")print(f"Unique passenger classes: {titanic['passenger_class'].nunique()}")print(f"Unique ages: {titanic['age'].nunique()}")
````

⚠️ **Warning:** unique() and nunique() exclude NaN values by default. For nunique(), use `dropna=False` to include them in the count.

✅ **Key Takeaway:** Use value_counts() to see distributions, unique() to see actual values, and nunique() when you only need the count.

## 5. Missing data handlingReal-world datasets almost always have missing values. Knowing how to detect, remove, and fill them is essential for data analysis.

### 5.1 Detecting missing values

📚 **Concept: isna() and notna()**These methods help identify missing data:- `.isna()` (or `.isnull()`) returns True for missing values- `.notna()` (or `.notnull()`) returns True for non-missing valuesThey work on both Series and DataFrames.

````python
# Count missing values per columntitanic.isna().sum()
````

````python
# Calculate percentage of missing valuesmissing_pct = (titanic.isna().sum() / len(titanic)) * 100missing_pct[missing_pct > 0]  # Show only columns with missing data
````

💡 **Tip:** The `age` column is missing ~20% of values—this is common and needs to be handled before analysis.

### 5.2 Removing missing data

📚 **Concept: The dropna() method**`.dropna()` removes rows or columns with missing values:- `axis='rows'` (default): drop rows with any NaN- `axis='columns'`: drop columns with any NaN- `how='any'` (default): drop if any NaN present- `how='all'`: drop only if all values are NaN- `thresh=n`: keep rows with at least n non-null values

````python
# Drop rows with any missing valuestitanic_complete = titanic.dropna()print(f"Original rows: {len(titanic)}")print(f"After dropna: {len(titanic_complete)}")
````

````python
# Drop rows that are completely emptytitanic_not_empty = titanic.dropna(how="all")print(f"Rows dropped: {len(titanic) - len(titanic_not_empty)}")
````

````python
# Keep rows with at least 7 non-null valuestitanic_thresh = titanic.dropna(thresh=7)print(f"Rows kept: {len(titanic_thresh)}")
````

⚠️ **Warning:** Dropping rows with dropna() can result in significant data loss. Consider the impact on your analysis before removing data!

### 5.3 Filling missing values

📚 **Concept: The fillna() method**`.fillna()` replaces missing values with specified values:- Fill with a constant: `fillna(0)`- Fill with a statistic: `fillna(df['col'].mean())`- Forward fill: `fillna(method='ffill')`- Backward fill: `fillna(method='bfill')`

````python
# Fill age with the median (more robust than mean for skewed data)median_age = titanic["age"].median()titanic["age"] = titanic["age"].fillna(median_age)# Verify no missing agesprint(f"Missing ages: {titanic['age'].isna().sum()}")
````

````python
# Fill embarked with the most common value (mode)most_common_port = titanic["embarked"].mode()[0]titanic["embarked"] = titanic["embarked"].fillna(most_common_port)print(f"Missing embarkation: {titanic['embarked'].isna().sum()}")
````

💡 **Tip:** Common filling strategies:- **Numerical data**: mean (if normally distributed) or median (if skewed)- **Categorical data**: mode (most frequent value) or a new category like "Unknown"- **Time series**: forward fill or backward fill

⚠️ **Warning:** Always use `.mode()[0]` not just `.mode()` when filling. The mode() method returns a Series, and we need to extract the first value.

✅ **Key Takeaway:** Filling is often better than dropping when you have significant missing data. Choose your fill strategy based on the data type and context.

## 6. Basic aggregationAggregation helps us summarize data and extract insights. We'll explore statistical methods and grouping operations.

### 6.1 Simple statistical aggregations

📚 **Concept: Built-in aggregation methods**Pandas provides many aggregation methods that work on Series and DataFrames:- `.mean()`, `.median()`: central tendency- `.min()`, `.max()`: extremes- `.sum()`: total- `.std()`, `.var()`: variability- `.count()`: non-null values

````python
# Average age of passengersprint(f"Mean age: {titanic['age'].mean():.1f}")print(f"Median age: {titanic['age'].median():.1f}")
````

````python
# Survival statisticssurvival_rate = titanic["survived"].mean()print(f"Overall survival rate: {survival_rate:.1%}")
````

````python
# Fare statisticsprint(f"Average fare: ${titanic['fare'].mean():.2f}")print(f"Median fare: ${titanic['fare'].median():.2f}")print(f"Max fare: ${titanic['fare'].max():.2f}")
````

💡 **Tip:** Notice how mean fare is higher than median fare? This suggests the data is right-skewed (a few very expensive tickets pull the average up).

### 6.2 Introduction to groupby

📚 **Concept: The groupby() method**`.groupby()` splits data into groups based on one or more columns, then applies aggregation functions to each group. This is one of the most powerful pandas operations!The pattern is: `df.groupby(column)[target_column].aggregation()`

````python
# Survival rate by passenger classtitanic.groupby("passenger_class")["survived"].mean()
````

````python
# Average age by passenger classtitanic.groupby("passenger_class")["age"].mean()
````

````python
# Multiple statistics at oncetitanic.groupby("passenger_class")["fare"].agg(["min", "median", "max"])
````

````python
# Group by multiple columns: class and sextitanic.groupby(["passenger_class", "sex"])["survived"].mean()
````

💡 **Tip:** The results show clear patterns—first class passengers and females had much higher survival rates!

✅ **Key Takeaway:** groupby() is essential for comparing groups and finding patterns in categorical data. Master this method!

---## Appendix: Advanced topicsThe following methods are useful for more advanced data manipulation. You don't need to master these immediately, but they're valuable tools to know about.

### A.1 Type conversion with astype()

📚 **Concept: Converting data types**The `.astype()` method converts columns to different data types. This is useful when pandas guesses the wrong type or when you need to optimize memory.

````python
# Convert passenger_class to string (categorical)titanic["passenger_class"] = titanic["passenger_class"].astype(str)# Check the resultprint(titanic["passenger_class"].dtype)
````

````python
# Convert survived to booleantitanic["survived"] = titanic["survived"].astype(bool)print(titanic["survived"].dtype)
````

💡 **Tip:** Use `astype('category')` for categorical variables with many repeating values—it can save significant memory.

### A.2 Custom transformations with apply() and map()

📚 **Concept: Applying custom functions**- `.apply()` applies a function to each element, row, or column- `.map()` maps values to other values (dictionary or function)These are powerful but can be slow on large datasets—vectorized operations are usually faster.

````python
# Create age categories using apply with a custom functiondef categorize_age(age):    if age < 13:        return "child"    elif age < 20:        return "teenager"    elif age < 60:        return "adult"    else:        return "senior"titanic["age_category"] = titanic["age"].apply(categorize_age)titanic["age_category"].value_counts()
````

````python
# Map embarkation codes to city namesport_mapping = {"C": "Cherbourg", "Q": "Queenstown", "S": "Southampton"}titanic["embark_city"] = titanic["embarked"].map(port_mapping)titanic[["embarked", "embark_city"]].head()
````

⚠️ **Warning:** apply() can be slow. For simple operations, vectorized alternatives are much faster.

### A.3 Binning data with cut() and qcut()

📚 **Concept: Creating categories from continuous data**- `pd.cut()` bins values into equal-width intervals- `pd.qcut()` bins values into equal-frequency intervals (quantiles)Both are useful for discretizing continuous variables.

````python
# Create age bins with equal widthsage_bins = pd.cut(titanic["age"], bins=5)age_bins.value_counts().sort_index()
````

````python
# Create fare quartiles (4 groups with equal counts)fare_quartiles = pd.qcut(titanic["fare"], q=4, labels=["Q1", "Q2", "Q3", "Q4"])fare_quartiles.value_counts()
````

💡 **Tip:** Use cut() when you want specific ranges (e.g., age groups 0-18, 18-35, 35+). Use qcut() when you want equally-sized groups.

### A.4 Correlation analysis

📚 **Concept: Measuring relationships**The `.corr()` method calculates correlation coefficients between numeric columns:- Values range from -1 (perfect negative) to +1 (perfect positive)- 0 means no linear relationship- Only works on numeric data

````python
# Calculate correlations between numeric columns# First convert survived back to int for correlationtitanic["survived"] = titanic["survived"].astype(int)numeric_cols = ["survived", "age", "siblings_spouses", "parents_children", "fare"]correlations = titanic[numeric_cols].corr()correlations
````

````python
# Look at correlations with survival specificallycorrelations["survived"].sort_values(ascending=False)
````

💡 **Tip:** The correlation shows that fare has a moderate positive relationship with survival—passengers who paid more were more likely to survive.

⚠️ **Warning:** Correlation measures linear relationships only. It doesn't imply causation!
