# Plotly Express code-along

**Goal (S7):** create clear, interactive visualisations programmatically that support a decision.

Install once: `pip install plotly nbformat kaleido`

```python
import pandas as pd
import plotly.express as px

gap = px.data.gapminder()
tips = px.data.tips()
```

## 1. The pattern
`px.<chart_type>(dataframe, x=..., y=..., color=..., ...)`

Columns are referred to **by name**. This is why tidy DataFrames matter.

## 2. Choosing the right chart
| Question | Chart |
|---|---|
| How does a value change over time? | line |
| How do categories compare? | bar |
| How is one variable distributed? | histogram, box |
| How are two numeric variables related? | scatter |
| How do parts make up a whole? | stacked bar (avoid pie charts) |

## 3. Scatter

```python
px.scatter(gap[gap["year"] == 2007], x="gdpPercap", y="lifeExp", size="pop", color="continent",
           hover_name="country", log_x=True, title="Wealth and health, 2007")
```

## 4. Line

```python
px.line(gap[gap["continent"] == "Europe"], x="year", y="lifeExp", color="country")
```

## 5. Bar

```python
avg = tips.groupby("day", as_index=False)["total_bill"].mean()
px.bar(avg, x="day", y="total_bill", text_auto=".1f", title="Average bill per day")
```

## 6. Distributions: histogram and box

```python
px.histogram(tips, x="total_bill", color="sex", barmode="overlay", nbins=25)
```

```python
px.box(tips, x="day", y="tip", color="smoker")
```

## 7. Facets: small multiples

```python
px.scatter(tips, x="total_bill", y="tip", facet_col="day", color="smoker", trendline="ols")
```

`trendline="ols"` needs `pip install statsmodels`.

## 8. Layout, titles and themes

```python
fig = px.line(gap[gap["country"].isin(["Sweden", "Germany", "China"])],
              x="year", y="gdpPercap", color="country", template="simple_white",
              title="GDP per capita", labels={"gdpPercap": "GDP per capita (USD)", "year": ""})
fig.update_layout(legend_title_text="", hovermode="x unified")
fig.update_traces(line_width=3)
fig
```

## 9. Interactivity: animation

```python
px.scatter(gap, x="gdpPercap", y="lifeExp", size="pop", color="continent", hover_name="country",
           animation_frame="year", log_x=True, range_y=[25, 90])
```

## 10. Exporting

```python
fig.write_html("gdp.html")      # interactive, shareable
# fig.write_image("gdp.png")    # needs: pip install kaleido
```

## 11. A reusable plotting function
Don't copy and paste chart code: wrap it in a function (the week 2 skills again).

```python
def line_chart(df, x, y, color=None, title=""):
    fig = px.line(df, x=x, y=y, color=color, title=title, template="simple_white")
    fig.update_layout(legend_title_text="")
    return fig

line_chart(gap[gap["country"].isin(["Sweden", "Norway"])], "year", "lifeExp", "country", "Life expectancy")
```
